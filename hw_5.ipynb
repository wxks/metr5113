```python
import sys
print(sys.executable)
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.ticker as tck
```

```python
x=np.arange(0,50,.1)
c= np.array([-2,-1,0,1,2])
def y_func(x_vals, c_val):
    return np.sqrt(2*x_vals) + c_val
plt.figure()
for i in range(len(c)):
    plt.plot(x,y_func(x, c[i]),label='c='+str(c[i]))
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.xlim(0,30)
plt.ylim(0,30)
plt.title('Question 3.2 Figure')
plt.legend()
plt.show()  

```
